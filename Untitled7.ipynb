```python
# ============================================================
# EXPLAINABLE AI (XAI) USING PRETRAINED RESNET-50
#
# Input : Uploaded girl photo
# Model : ResNet-50 pretrained on ImageNet
#
# XAI METHODS:
# 1. Occlusion
# 2. Integrated Gradients
# 3. Guided Backpropagation
# 4. Grad-CAM
# 5. Grad-CAM++
# 6. LIME
# 7. SHAP
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install -q captum shap lime scikit-image grad-cam


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import torch
import torch.nn as nn

import numpy as np
import matplotlib.pyplot as plt

from PIL import Image

import torchvision.transforms as transforms
from torchvision.models import resnet50, ResNet50_Weights

# -----------------------------
# Captum
# -----------------------------
from captum.attr import (
    IntegratedGradients,
    Occlusion,
    GuidedBackprop
)

# -----------------------------
# Grad-CAM
# -----------------------------
from pytorch_grad_cam import (
    GradCAM,
    GradCAMPlusPlus
)

from pytorch_grad_cam.utils.image import (
    show_cam_on_image
)

from pytorch_grad_cam.utils.model_targets import (
    ClassifierOutputTarget
)

# -----------------------------
# LIME
# -----------------------------
from lime import lime_image

from skimage.segmentation import (
    mark_boundaries
)

# -----------------------------
# SHAP
# -----------------------------
import shap

# -----------------------------
# Google Colab upload
# -----------------------------
from google.colab import files


# ============================================================
# 3. SELECT DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 60)
print("DEVICE")
print("=" * 60)

print("Using device:", device)


# ============================================================
# 4. LOAD PRETRAINED RESNET-50
# ============================================================

print()
print("=" * 60)
print("LOADING RESNET-50")
print("=" * 60)

weights = ResNet50_Weights.DEFAULT

model = resnet50(
    weights=weights
)


# ------------------------------------------------------------
# IMPORTANT:
# Disable in-place ReLU operations.
#
# This helps Captum and Grad-CAM work safely with ResNet-50.
# ------------------------------------------------------------

for module in model.modules():

    if isinstance(module, nn.ReLU):

        module.inplace = False


model = model.eval()

model = model.to(device)


# ImageNet preprocessing
preprocess = weights.transforms()


# ImageNet class names
imagenet_labels = weights.meta["categories"]


print("ResNet-50 loaded successfully.")

print(
    "Number of ImageNet classes:",
    len(imagenet_labels)
)


# ============================================================
# 5. UPLOAD YOUR GIRL PHOTO
# ============================================================

print()
print("=" * 60)
print("UPLOAD IMAGE")
print("=" * 60)

print("Please upload your girl photo.")

uploaded = files.upload()


# Get uploaded filename
image_path = list(
    uploaded.keys()
)[0]


# Open image
img_pil = Image.open(
    image_path
).convert("RGB")


print(
    "Image loaded:",
    image_path
)

print(
    "Original image size:",
    img_pil.size
)


# ============================================================
# 6. DISPLAY ORIGINAL IMAGE
# ============================================================

plt.figure(
    figsize=(6, 5)
)

plt.imshow(
    img_pil
)

plt.title(
    "Original Uploaded Image"
)

plt.axis("off")

plt.show()


# ============================================================
# 7. PREPARE IMAGE FOR RESNET-50
# ============================================================

def prepare_image(image):

    tensor = preprocess(
        image
    )

    tensor = tensor.unsqueeze(0)

    tensor = tensor.to(device)

    return tensor


input_tensor = prepare_image(
    img_pil
)


print()
print(
    "Input tensor shape:",
    input_tensor.shape
)


# ============================================================
# 8. RESNET-50 PREDICTION
# ============================================================

def predict_tensor(tensor):

    model.eval()

    with torch.no_grad():

        logits = model(
            tensor
        )

        probabilities = torch.softmax(
            logits,
            dim=1
        )

    return probabilities, logits


probs, logits = predict_tensor(
    input_tensor
)


# Highest probability
top_prob, top_idx = probs.max(
    dim=1
)


# Target class
target = top_idx.item()


# Predicted ImageNet label
pred_label = imagenet_labels[
    target
]


print()
print("=" * 60)
print("RESNET-50 PREDICTION")
print("=" * 60)

print(
    "Predicted class :",
    pred_label
)

print(
    "Class ID :",
    target
)

print(
    "Probability :",
    round(
        top_prob.item(),
        4
    )
)

print(
    "Confidence :",
    round(
        top_prob.item() * 100,
        2
    ),
    "%"
)


# ============================================================
# 9. OCCLUSION
# ============================================================

print()
print("=" * 60)
print("1. OCCLUSION")
print("=" * 60)

occlusion = Occlusion(
    model
)


attributions_occ = occlusion.attribute(

    input_tensor,

    strides=(1, 8, 8),

    target=target,

    sliding_window_shapes=(1, 15, 15),

    baselines=0
)


# Convert to NumPy
attr_occ = (

    attributions_occ

    .squeeze()

    .detach()

    .cpu()

    .numpy()
)


# Combine RGB channels
attr_occ = attr_occ.sum(
    axis=0
)


# Normalize
attr_occ_norm = (

    attr_occ - attr_occ.min()

) / (

    attr_occ.max()
    - attr_occ.min()
    + 1e-8
)


# Display
plt.figure(
    figsize=(10, 4)
)


plt.subplot(
    1, 2, 1
)

plt.imshow(
    img_pil
)

plt.title(
    "Original"
)

plt.axis("off")


plt.subplot(
    1, 2, 2
)

plt.imshow(
    img_pil
)

plt.imshow(
    attr_occ_norm,
    cmap="jet",
    alpha=0.5
)

plt.title(
    "Occlusion Sensitivity"
)

plt.axis("off")


plt.tight_layout()

plt.show()


print(
    "Occlusion completed successfully."
)


# ============================================================
# 10. INTEGRATED GRADIENTS
# ============================================================

print()
print("=" * 60)
print("2. INTEGRATED GRADIENTS")
print("=" * 60)


ig = IntegratedGradients(
    model
)


# Zero baseline
baseline = torch.zeros_like(
    input_tensor
).to(device)


attributions_ig = ig.attribute(

    input_tensor,

    baselines=baseline,

    target=target,

    n_steps=50
)


# Convert to NumPy
attr_ig = (

    attributions_ig

    .squeeze()

    .detach()

    .cpu()

    .numpy()
)


# Combine RGB channels
attr_ig = attr_ig.sum(
    axis=0
)


# Normalize
attr_ig_norm = (

    attr_ig - attr_ig.min()

) / (

    attr_ig.max()
    - attr_ig.min()
    + 1e-8
)


# Display
plt.figure(
    figsize=(10, 4)
)


plt.subplot(
    1, 2, 1
)

plt.imshow(
    img_pil
)

plt.title(
    "Original"
)

plt.axis("off")


plt.subplot(
    1, 2, 2
)

plt.imshow(
    img_pil
)

plt.imshow(
    attr_ig_norm,
    cmap="hot",
    alpha=0.5
)

plt.title(
    "Integrated Gradients"
)

plt.axis("off")


plt.tight_layout()

plt.show()


print(
    "Integrated Gradients completed successfully."
)


# ============================================================
# 11. GUIDED BACKPROPAGATION
# ============================================================

print()
print("=" * 60)
print("3. GUIDED BACKPROPAGATION")
print("=" * 60)


gbp = GuidedBackprop(
    model
)


# Create fresh input
gbp_input = (
    input_tensor
    .detach()
    .clone()
)


gbp_input.requires_grad = True


attributions_gbp = gbp.attribute(

    gbp_input,

    target=target
)


# Convert to NumPy
attr_gbp = (

    attributions_gbp

    .squeeze()

    .detach()

    .cpu()

    .numpy()
)


# Absolute values
attr_gbp = np.abs(
    attr_gbp
)


# Combine RGB channels
attr_gbp = attr_gbp.sum(
    axis=0
)


# Normalize
attr_gbp_norm = (

    attr_gbp - attr_gbp.min()

) / (

    attr_gbp.max()
    - attr_gbp.min()
    + 1e-8
)


# Display
plt.figure(
    figsize=(10, 4)
)


plt.subplot(
    1, 2, 1
)

plt.imshow(
    img_pil
)

plt.title(
    "Original"
)

plt.axis("off")


plt.subplot(
    1, 2, 2
)

plt.imshow(
    attr_gbp_norm,
    cmap="gray"
)

plt.title(
    "Guided Backpropagation"
)

plt.axis("off")


plt.tight_layout()

plt.show()


print(
    "Guided Backpropagation completed successfully."
)


# ============================================================
# 12. GRAD-CAM
# ============================================================

print()
print("=" * 60)
print("4. GRAD-CAM")
print("=" * 60)


model.eval()


# Last convolutional layer
target_layers = [
    model.layer4[-1]
]


cam = GradCAM(

    model=model,

    target_layers=target_layers
)


targets = [
    ClassifierOutputTarget(
        target
    )
]


grayscale_cam = cam(

    input_tensor=input_tensor,

    targets=targets
)[0]


# Resize image
H_cam, W_cam = (
    grayscale_cam.shape
)


img_resized = img_pil.resize(
    (W_cam, H_cam)
)


# Convert image to 0-1
rgb_img = (

    np.array(
        img_resized
    )

    .astype(
        np.float32
    )

    / 255.0
)


# Create overlay
visualization = show_cam_on_image(

    rgb_img,

    grayscale_cam,

    use_rgb=True
)


# Display
plt.figure(
    figsize=(10, 4)
)


plt.subplot(
    1, 2, 1
)

plt.imshow(
    img_pil
)

plt.title(
    "Original\n" + pred_label
)

plt.axis("off")


plt.subplot(
    1, 2, 2
)

plt.imshow(
    visualization
)

plt.title(
    "Grad-CAM"
)

plt.axis("off")


plt.tight_layout()

plt.show()


print(
    "Grad-CAM completed successfully."
)


# ============================================================
# 13. GRAD-CAM++
# ============================================================

print()
print("=" * 60)
print("5. GRAD-CAM++")
print("=" * 60)


target_layers = [
    model.layer4[-1]
]


cam_pp = GradCAMPlusPlus(

    model=model,

    target_layers=target_layers
)


targets = [
    ClassifierOutputTarget(
        target
    )
]


grayscale_cam_pp = cam_pp(

    input_tensor=input_tensor,

    targets=targets
)[0]


# Resize image
H_cam_pp, W_cam_pp = (
    grayscale_cam_pp.shape
)


img_resized_pp = img_pil.resize(
    (W_cam_pp, H_cam_pp)
)


# Convert to 0-1
rgb_img_pp = (

    np.array(
        img_resized_pp
    )

    .astype(
        np.float32
    )

    / 255.0
)


# Create overlay
visualization_pp = show_cam_on_image(

    rgb_img_pp,

    grayscale_cam_pp,

    use_rgb=True
)


# Display
plt.figure(
    figsize=(10, 4)
)


plt.subplot(
    1, 2, 1
)

plt.imshow(
    img_pil
)

plt.title(
    "Original\n" + pred_label
)

plt.axis("off")


plt.subplot(
    1, 2, 2
)

plt.imshow(
    visualization_pp
)

plt.title(
    "Grad-CAM++"
)

plt.axis("off")


plt.tight_layout()

plt.show()


print(
    "Grad-CAM++ completed successfully."
)


# ============================================================
# 14. LIME
# ============================================================

print()
print("=" * 60)
print("6. LIME")
print("=" * 60)


# ------------------------------------------------------------
# LIME prediction function
# ------------------------------------------------------------

def batch_predict(images):

    model.eval()

    batch_tensors = []


    for image in images:

        # Ensure values 0-1
        image = np.clip(
            image,
            0,
            1
        )


        # NumPy -> PIL
        pil_image = Image.fromarray(

            (
                image * 255
            ).astype(
                np.uint8
            )

        )


        # ResNet preprocessing
        tensor = preprocess(
            pil_image
        )


        batch_tensors.append(
            tensor
        )


    # Stack images
    batch = torch.stack(
        batch_tensors
    ).to(device)


    # Prediction
    with torch.no_grad():

        logits = model(
            batch
        )

        probabilities = torch.softmax(
            logits,
            dim=1
        )


    return probabilities.cpu().numpy()


# ------------------------------------------------------------
# Convert image to H x W x 3
# ------------------------------------------------------------

img_lime = (

    np.array(
        img_pil
    )

    .astype(
        np.float32
    )

    / 255.0
)


# ------------------------------------------------------------
# Create LIME explainer
# ------------------------------------------------------------

lime_explainer = (
    lime_image.LimeImageExplainer()
)


# ------------------------------------------------------------
# Generate explanation
# ------------------------------------------------------------

lime_explanation = (

    lime_explainer.explain_instance(

        img_lime,

        batch_predict,

        top_labels=5,

        hide_color=0,

        num_samples=100

    )
)


# ------------------------------------------------------------
# Get explanation for predicted class
# ------------------------------------------------------------

temp, mask = (

    lime_explanation.get_image_and_mask(

        target,

        positive_only=True,

        num_features=5,

        hide_rest=False

    )
)


# ------------------------------------------------------------
# Display LIME
# ------------------------------------------------------------

plt.figure(
    figsize=(10, 4)
)


plt.subplot(
    1, 2, 1
)

plt.imshow(
    img_pil
)

plt.title(
    "Original\n" + pred_label
)

plt.axis("off")


plt.subplot(
    1, 2, 2
)

plt.imshow(

    mark_boundaries(

        temp / 255.0,

        mask

    )

)

plt.title(
    "LIME - Important Superpixels"
)

plt.axis("off")


plt.tight_layout()

plt.show()


print(
    "LIME completed successfully."
)


# ============================================================
# 15. SHAP
# ============================================================
#
# IMPORTANT:
#
# We use SHAP Partition Explainer.
#
# We DO NOT use:
#
# shap.DeepExplainer
#
# or:
#
# shap.GradientExplainer
#
# because these can produce the ResNet-50
# "view + inplace" backward error.
# ============================================================

print()
print("=" * 60)
print("7. SHAP")
print("=" * 60)


# ------------------------------------------------------------
# 15.1 Resize image to 224 x 224
# ------------------------------------------------------------

shap_image = img_pil.resize(
    (224, 224)
)


# Convert to NumPy [0,1]
shap_image_np = (

    np.array(
        shap_image
    )

    .astype(
        np.float32
    )

    / 255.0
)


print(
    "SHAP image shape:",
    shap_image_np.shape
)


# ------------------------------------------------------------
# IMPORTANT FIX
#
# ImageMasker expects:
#
# (224,224,3)
#
# But Explainer input should be:
#
# (1,224,224,3)
# ------------------------------------------------------------

shap_images = np.expand_dims(

    shap_image_np,

    axis=0

)


print(
    "SHAP batch shape:",
    shap_images.shape
)


# ============================================================
# 15.2 SHAP prediction function
# ============================================================

def shap_predict(images):

    model.eval()

    batch_tensors = []


    for image in images:

        # Make sure values are 0-1
        image = np.clip(
            image,
            0,
            1
        )


        # NumPy -> PIL
        pil_image = Image.fromarray(

            (
                image * 255
            ).astype(
                np.uint8
            )

        )


        # ResNet preprocessing
        tensor = preprocess(
            pil_image
        )


        batch_tensors.append(
            tensor
        )


    # Create PyTorch batch
    batch = torch.stack(
        batch_tensors
    ).to(device)


    # Model prediction
    with torch.no_grad():

        output = model(
            batch
        )

        probabilities = torch.softmax(
            output,
            dim=1
        )


    return probabilities.cpu().numpy()


# ============================================================
# 15.3 TEST SHAP PREDICTION
# ============================================================

test_prediction = shap_predict(
    shap_images
)


print(
    "SHAP prediction shape:",
    test_prediction.shape
)


# ============================================================
# 15.4 CREATE SHAP IMAGE MASKER
# ============================================================

masker = shap.maskers.Image(

    "blur(32,32)",

    shap_image_np.shape

)


print(
    "SHAP ImageMasker created successfully."
)


# ============================================================
# 15.5 CREATE PARTITION EXPLAINER
# ============================================================

shap_explainer = shap.Explainer(

    shap_predict,

    masker,

    algorithm="partition",

    output_names=imagenet_labels

)


print(
    "SHAP Partition Explainer created successfully."
)


# ============================================================
# 15.6 CALCULATE SHAP
# ============================================================

print()
print(
    "Calculating SHAP values..."
)

print(
    "Please wait..."
)


shap_result = shap_explainer(

    shap_images,

    max_evals=300,

    batch_size=16,

    outputs=[target]

)


print()
print(
    "SHAP calculation completed."
)


# ============================================================
# 15.7 GET SHAP VALUES
# ============================================================

shap_values = shap_result.values


print(
    "Raw SHAP shape:",
    shap_values.shape
)


# ============================================================
# 15.8 REMOVE BATCH DIMENSION
# ============================================================

# Usually:
#
# (1,224,224,3,1)
#
# becomes:
#
# (224,224,3,1)

if shap_values.ndim == 5:

    shap_values = shap_values[0]


# Remove output dimension
#
# (224,224,3,1)
#
# becomes:
#
# (224,224,3)

if shap_values.ndim == 4:

    if shap_values.shape[-1] == 1:

        shap_values = (
            shap_values[..., 0]
        )


print(
    "Processed SHAP shape:",
    shap_values.shape
)


# ============================================================
# 15.9 COMBINE RGB CHANNELS
# ============================================================

if shap_values.ndim == 3:

    shap_map = np.mean(

        np.abs(
            shap_values
        ),

        axis=-1

    )

else:

    shap_map = np.abs(
        shap_values
    )


# ============================================================
# 15.10 NORMALIZE SHAP MAP
# ============================================================

max_value = np.max(
    np.abs(shap_map)
)


if max_value > 0:

    shap_norm = (

        shap_map
        / max_value

    )

else:

    shap_norm = shap_map


# ============================================================
# 15.11 DISPLAY SHAP
# ============================================================

plt.figure(
    figsize=(15, 5)
)


# -----------------------------
# Original
# -----------------------------

plt.subplot(
    1, 3, 1
)

plt.imshow(
    shap_image_np
)

plt.title(
    "Original\n" + pred_label
)

plt.axis("off")


# -----------------------------
# SHAP map
# -----------------------------

plt.subplot(
    1, 3, 2
)

plt.imshow(

    shap_norm,

    cmap="hot"

)

plt.colorbar()

plt.title(
    "SHAP Importance Map"
)

plt.axis("off")


# -----------------------------
# SHAP overlay
# -----------------------------

plt.subplot(
    1, 3, 3
)

plt.imshow(
    shap_image_np
)

plt.imshow(

    shap_norm,

    cmap="jet",

    alpha=0.5

)

plt.title(
    "SHAP Overlay"
)
plt.axis("off")


plt.tight_layout()

plt.show()


print()
print(
    "SHAP completed successfully."
)


# ============================================================
# 16. FINAL SUMMARY
# ============================================================

print()
print("=" * 60)

print(
    " XAI ANALYSIS COMPLETED"
)

print("=" * 60)


print()

print(
    "Input Image :",
    image_path
)

print(
    "Predicted Class :",
    pred_label
)

print(
    "Class ID :",
    target
)

print(
    "Confidence :",
    round(
        top_prob.item() * 100,
        2
    ),
    "%"
)


print()
print(
    "Methods Applied:"
)

print(
    "1. Occlusion ✓"
)

print(
    "2. Integrated Gradients ✓"
)

print(
    "3. Guided Backpropagation ✓"
)

print(
    "4. Grad-CAM ✓"
)

print(
    "5. Grad-CAM++ ✓"
)

print(
    "6. LIME ✓"
)

print(
    "7. SHAP ✓"
)


print()
print(
    "All XAI methods completed successfully."
)

print("=" * 60)
```
